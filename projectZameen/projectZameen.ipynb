```python
import re

def convert_into_crore(text):

    if text is None or not isinstance(text, str):
        return None

    match = re.search(r"PKR\s*(\d+(?:\.\d+)?)\s*([a-zA-Z]+)", text)

    if not match:
        return None

    value = float(match.group(1))
    unit = match.group(2).lower()

    if unit == "crore":
        return round(value * 100, 2)

    elif unit == "lakh":
        return value

    elif unit == "arab":
        return round(value * 10000, 2)

    return None
```

```python
# df['price'] = df['price'].apply(convert_into_crore)
df.insert(2,"price_lakh",df.price.apply(convert_into_crore))
```

```python
def convert_area(area):

    match = re.search(
        r'(\d+(?:\.\d+)?)\s*(Marla|Kanal|Sq\. Yd\.)',
        str(area),
        re.IGNORECASE
    )

    if not match:
        return None

    value = float(match.group(1))
    unit = match.group(2).lower()

    if unit == 'kanal':
        return value * 20

    elif unit == 'sq. yd.':
        return (value * 9) / 225

    elif unit == 'marla':
        return value

    return None


df.insert(7,"area_value",df.area.apply(convert_area))
```
