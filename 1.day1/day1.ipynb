Level 3 — ML-oriented Challenge
You are given a list of feature values representing house sizes in square feet: house_sizes = [850, 900, 1200, 1500, 2000, 750, 1100]. Print:
The number of houses
The smallest and largest house
The average house size (rounded to 2 decimal places using an f-string)
You are given student_record = ["Ayesha", 21, [90, 87, 93]] (a mixed record: name, age, marks list). Print a formatted sentence using f-strings that includes the student's name, age, and their average mark.
Given a raw text feature raw = " Income: 55000 ", extract just the numeric part as an integer (hint: use .split(":"), .strip(), and int()).

```python
student_record = ["Ayesha", 21, [90, 87, 93]]

# Unpack data and calculate average mark
name = student_record[0]
age = student_record[1]
marks = student_record[2]
average_mark = sum(marks) / len(marks)

# Print formatted sentence
print(f"{name} is {age} years old and has an average mark of {average_mark:.2f}.")
```

```python
raw = "  Income: 55000  "

# Split by colon, grab the second part, strip whitespace, and convert to integer
income_val = int(raw.split(":")[1].strip())

print(income_val)
print(type(income_val))

```
