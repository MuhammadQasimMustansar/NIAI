```python
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import MinMaxScaler
from sklearn.pipeline import Pipeline,make_pipeline
from sklearn.feature_selection import SelectKBest,chi2
from sklearn.tree import DecisionTreeClassifier
```

```python
df = sns.load_dataset("titanic")
df.head()
```

```python
df = df[['pclass', 'sex', 'age', 'sibsp', 'parch', 'fare', 'embarked', 'survived']]
df.head()
```

```python
df.info()
```

```python
df.shape
```

```python
df.isnull().sum()
```

```python
X_train,X_test,y_train,y_test = train_test_split(df.drop(columns=['survived']), df['survived'], test_size=0.2, random_state=42)
```

```python
type(X_train)
```

```python
X_train.head()
```

```python
y_train.head()
```

```python
trf1 = ColumnTransformer([
    ('impute_age', SimpleImputer(), [2]),
    ('imputeembarked',SimpleImputer(strategy='most_frequent'), [6])
], remainder='passthrough')
```

```python
# X_train_processed = trf1.fit_transform(X_train)
```

```python
# X_test_processed = trf1.transform(X_test)
```

```python
type(X_train)
```

```python
# X_train.isnull().sum()
```

```python
trf2 = ColumnTransformer([
    ('ohe_sex_embarked', OneHotEncoder(sparse_output=False, handle_unknown='ignore'), [1,6])
], remainder='passthrough')
```

```python
X_train.shape
```

```python
trf3 = ColumnTransformer([
    ('sacle', MinMaxScaler(), slice(0, 7))
])
```

```python
trf4 = SelectKBest(score_func=chi2, k=8)
```

```python
trf5 = DecisionTreeClassifier()
```

```python
pipe = Pipeline([
    ('trf1',trf1),
    ('trf2',trf2),
    ('trf3',trf3),
    ('trf4',trf4),
    ('trf5',trf5)
])
pipe
```

```python
# pipe = make_pipeline(trf1,trf2,trf3,trf4,trf5)
```

```python
# pipe.fit_transform(X_train, y_train) use fit_transform when we not train a model but when we train (decision tree) then fit and then predict

pipe.fit(X_train, y_train)
```

```python
pipe.named_steps
```

```python
pipe.named_steps['trf1'].transformers_[1][1].statistics_
```

```python
pipe.named_steps['trf2'].transformers_[0][1].n_features_in_
```

```python
y_pred = pipe.predict(X_test)
y_pred
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test, y_pred)
```
